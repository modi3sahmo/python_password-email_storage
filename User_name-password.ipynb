```python
spcs = ['!', '@', '#', '$', '%', '&', '*']
types = ['yahoo', 'gmail', 'hotmail']
user_data ={}
def Email():
    name = input("Enter your name: ")
    while True:
        type_ = input("Choose provider (yahoo/gmail/hotmail): ").lower()
        if type_ in types:
            email = f"{name}@{type_}.com"
            return email  
        else:
            print("Invalid provider! Please choose from yahoo, gmail, or hotmail.")

def password():
    while True:
        pwd = input("Enter password: ")

        if len(pwd) == 12:
            has_spcs = False
            has_digit =False
            for char in pwd:
                if char in spcs:
                    has_spcs = True
                if char.isdigit():
                    has_digit = True    
                    break 
            if has_spcs and has_digit:
                    print("Valid Password!")
                    return pwd
            elif not has_spcs:
                print("Invalid Password! Must contain at least one special character.")
            elif not has_digit:
                print("Invalid Password! Must contain at least one numeric digit.")
        elif len(pwd) > 12:
            print(f"'{pwd}' is too long, enter 12 characters only.")
        else:
            print("Invalid Password! Must be exactly 12 characters.") 
def record():
    while True:
        username =input("Please Enter the user: ")
        if username in user_data:
            print("Duplicate user found!!")
        else:
            email_id =Email()
            pwd =password()
            user_data[username] = {"Email": email_id, "Password": pwd}
            print("User data saved successfully!")
            print(user_data) 

        another = input("Do you want another user(yes/no)? ").lower()
        if another != "yes":
            break
        

record()
```
